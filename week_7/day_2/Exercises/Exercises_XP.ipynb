## Exercise 2: Artificial Neural Networks (ANNs) Architecture

```python
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from matplotlib.patches import FancyBboxPatch, FancyArrowPatch

# Create ANN diagram
fig, ax = plt.subplots(figsize=(14, 8))
ax.set_xlim(0, 10)
ax.set_ylim(0, 10)
ax.axis('off')

# Define positions
input_y = [8, 5, 2]
hidden_y = [8.5, 6.5, 4.5, 2.5]
output_y = [6.5, 3.5]

# Draw neurons
neuron_size = 0.3

# Input layer
for i, y in enumerate(input_y):
    circle = plt.Circle((1, y), neuron_size, color='#FF6B6B', ec='black', linewidth=2, zorder=3)
    ax.add_patch(circle)
    ax.text(1, y, f'x{i+1}', ha='center', va='center', fontsize=10, fontweight='bold', color='white')
    ax.text(0.3, y, f'Input {i+1}', ha='right', va='center', fontsize=9)

# Hidden layer
for j, y in enumerate(hidden_y):
    circle = plt.Circle((5, y), neuron_size, color='#4ECDC4', ec='black', linewidth=2, zorder=3)
    ax.add_patch(circle)
    ax.text(5, y, f'h{j+1}', ha='center', va='center', fontsize=10, fontweight='bold', color='white')

# Output layer
for k, y in enumerate(output_y):
    circle = plt.Circle((9, y), neuron_size, color='#95E1D3', ec='black', linewidth=2, zorder=3)
    ax.add_patch(circle)
    ax.text(9, y, f'y{k+1}', ha='center', va='center', fontsize=10, fontweight='bold', color='white')
    ax.text(9.7, y, f'Output {k+1}', ha='left', va='center', fontsize=9)

# Draw connections from input to hidden
for i, y1 in enumerate(input_y):
    for j, y2 in enumerate(hidden_y):
        arrow = FancyArrowPatch((1.3, y1), (4.7, y2), 
                               arrowstyle='->', mutation_scale=15, 
                               color='gray', alpha=0.5, linewidth=1.5, zorder=1)
        ax.add_patch(arrow)
        # Add weight labels for first few connections
        if i == 0 and j == 0:
            mid_x, mid_y = (1.3 + 4.7) / 2, (y1 + y2) / 2
            ax.text(mid_x, mid_y + 0.3, f'w[{i},{j}]', fontsize=8, style='italic', color='darkblue')

# Draw connections from hidden to output
for j, y2 in enumerate(hidden_y):
    for k, y3 in enumerate(output_y):
        arrow = FancyArrowPatch((5.3, y2), (8.7, y3), 
                               arrowstyle='->', mutation_scale=15, 
                               color='gray', alpha=0.5, linewidth=1.5, zorder=1)
        ax.add_patch(arrow)

# Add layer labels
ax.text(1, 9.5, 'INPUT LAYER\n(3 neurons)', ha='center', fontsize=11, fontweight='bold', 
        bbox=dict(boxstyle='round', facecolor='#FFE5E5', alpha=0.7))
ax.text(5, 9.5, 'HIDDEN LAYER\n(4 neurons)', ha='center', fontsize=11, fontweight='bold',
        bbox=dict(boxstyle='round', facecolor='#E0F7F6', alpha=0.7))
ax.text(9, 9.5, 'OUTPUT LAYER\n(2 neurons)', ha='center', fontsize=11, fontweight='bold',
        bbox=dict(boxstyle='round', facecolor='#E5F9F7', alpha=0.7))

# Add annotations for components
ax.text(0.5, 0.5, 'COMPONENTS LABELED:', fontsize=10, fontweight='bold')
ax.text(0.5, 0.1, '• Neurons (circles): Process information\n• Weights w[i,j]: Scale each connection\n• Biases: Shift activation thresholds\n• Activation: Non-linear transformation (ReLU/Softmax)',
        fontsize=9, family='monospace')

plt.tight_layout()
plt.title('Artificial Neural Network Architecture (3-4-2)', fontsize=14, fontweight='bold', pad=20)
plt.show()

print("\nANN DIAGRAM GENERATED")
print("="*80)
```

### Forward Propagation Explanation

```python
print("\nFORWARD PROPAGATION THROUGH THE NETWORK")
print("="*80)
print("""
Forward propagation is the process by which information flows from input to output 
through the network. Starting with the input layer, each neuron receives raw features 
(x1, x2, x3). These inputs are multiplied by corresponding weights w[i,j], summing 
to create a weighted sum for each hidden neuron: z = Σ(w[i,j] * x[i]) + b[j], where 
b[j] is the bias term. The bias allows the network to shift the decision boundary 
independently of the inputs. Next, an activation function (e.g., ReLU) is applied to 
z, introducing non-linearity: h[j] = ReLU(z[j]). This non-linear transformation 
enables the network to learn complex patterns beyond linear combinations. Finally, 
the hidden layer outputs (h1, h2, h3, h4) become inputs to the output layer, where 
the same weighted sum and activation process repeats: output = activation(W_hidden_to_output × h + b_output). 
The final output neurons produce predictions (y1, y2) based on all hierarchical 
processing above them. This entire forward pass is differentiable, enabling training 
via backpropagation.
""")
```

## Exercise 3: Creating and Visualizing the Dataset

```python
# PREFILLED: Import and set seed
np.random.seed(0)
x = np.arange(-1, 1, 0.1)  # 20 points from -1 to <1 step 0.1

print(f"x shape: {x.shape}")
print(f"x values: {x}")
```

```python
# SOLUTION: Generate y with noise from y = -x^2 + Gaussian noise N(0, 0.05)
y_true = -x**2  # True underlying function
noise = np.random.normal(0, 0.05, size=x.shape)  # Gaussian noise
y = y_true + noise  # Add noise to true function

print(f"y shape: {y.shape}")
print(f"\nTrue values (y = -x^2):")
print(y_true)
print(f"\nNoisy values (y = -x^2 + noise):")
print(y)
print(f"\nNoise statistics:")
print(f"  Mean: {noise.mean():.6f}")
print(f"  Std Dev: {noise.std():.6f}")
```

```python
# PREFILLED: Scatter plot of noisy dataset
plt.figure(figsize=(8, 5))
plt.scatter(x, y, color='#FF6B6B', s=100, alpha=0.7, edgecolors='black', linewidth=1.5, label='Noisy Data')
plt.plot(x, y_true, 'b--', linewidth=2, alpha=0.6, label='True Function: y = -x²')
plt.xlabel('x', fontsize=12, fontweight='bold')
plt.ylabel('y', fontsize=12, fontweight='bold')
plt.title('Noisy Dataset Sampled from y = -x²', fontsize=13, fontweight='bold')
plt.grid(True, alpha=0.3)
plt.legend(fontsize=10)
plt.tight_layout()
plt.show()

print("Dataset visualization complete!")
```

```python
# SOLUTION: Split into train (first 12) and test (last 8) sets
x_train = x[:12]
y_train = y[:12]

x_test = x[12:]
y_test = y[12:]

print(f"Train sizes: x_train shape = {x_train.shape}, y_train shape = {y_train.shape}")
print(f"Test sizes:  x_test shape = {x_test.shape}, y_test shape = {y_test.shape}")
print(f"\nTrain split (first 12 points):")
print(f"  x_train: {x_train}")
print(f"  y_train: {y_train}")
print(f"\nTest split (last 8 points):")
print(f"  x_test: {x_test}")
print(f"  y_test: {y_test}")
```

```python
# Visualize train/test split
plt.figure(figsize=(8, 5))
plt.scatter(x_train, y_train, color='#4ECDC4', s=120, alpha=0.8, edgecolors='black', linewidth=1.5, label='Train Set (12 points)', marker='o')
plt.scatter(x_test, y_test, color='#FF6B6B', s=120, alpha=0.8, edgecolors='black', linewidth=1.5, label='Test Set (8 points)', marker='^')
plt.plot(x, y_true, 'b--', linewidth=2, alpha=0.5, label='True Function: y = -x²')
plt.xlabel('x', fontsize=12, fontweight='bold')
plt.ylabel('y', fontsize=12, fontweight='bold')
plt.title('Train/Test Split (12 train, 8 test)', fontsize=13, fontweight='bold')
plt.grid(True, alpha=0.3)
plt.legend(fontsize=10)
plt.tight_layout()
plt.show()

print("Train/test visualization complete!")
```

## Exercise 4: Fitting Polynomial Models of Different Degrees

```python
# SOLUTION: Define polynomial_fit function
def polynomial_fit(degree: int):
    """
    Fit a polynomial of given degree to the training data.
    
    Parameters:
        degree: Degree of the polynomial
    
    Returns:
        np.poly1d: Callable polynomial object
    """
    # Use np.polyfit to get coefficients from training data
    coeffs = np.polyfit(x_train, y_train, degree)
    # Return polynomial as callable object
    return np.poly1d(coeffs)

# Test the function
print("Testing polynomial_fit function:")
for deg in [1, 2, 3]:
    poly = polynomial_fit(deg)
    print(f"\nDegree {deg}: {poly}")
```

```python
# SOLUTION: Define plot_polyfit function
def plot_polyfit(degree: int):
    """
    Plot polynomial fit with training and test data.
    
    Parameters:
        degree: Degree of the polynomial to fit and plot
    """
    # Step 1: Fit polynomial to training data
    poly = polynomial_fit(degree)
    
    # Step 2: Create dense evaluation points
    x_plot = np.linspace(x.min() - 0.1, x.max() + 0.1, 200)
    
    # Step 3: Evaluate polynomial on dense points
    y_plot = poly(x_plot)
    
    # Step 4: Create figure
    plt.figure(figsize=(10, 6))
    
    # Scatter train and test points
    plt.scatter(x_train, y_train, color='#4ECDC4', s=100, alpha=0.8, 
               edgecolors='black', linewidth=1.5, label='Train Set', marker='o', zorder=3)
    plt.scatter(x_test, y_test, color='#FF6B6B', s=100, alpha=0.8, 
               edgecolors='black', linewidth=1.5, label='Test Set', marker='^', zorder=3)
    
    # Plot fitted polynomial curve
    plt.plot(x_plot, y_plot, 'purple', linewidth=2.5, label=f'Polynomial Fit (Degree {degree})', zorder=2)
    
    # Plot true function for reference
    plt.plot(x, y_true, 'b--', linewidth=2, alpha=0.5, label='True Function: y = -x²', zorder=1)
    
    # Formatting
    plt.xlabel('x', fontsize=12, fontweight='bold')
    plt.ylabel('y', fontsize=12, fontweight='bold')
    plt.title(f'Polynomial Regression Fit (Degree {degree})', fontsize=13, fontweight='bold')
    plt.grid(True, alpha=0.3)
    plt.legend(fontsize=10, loc='best')
    plt.tight_layout()
    
    # Step 5: Display
    plt.show()

# SOLUTION: Visualize degrees 1, 7, and 11
print("\nVisualizing polynomial fits for degrees 1, 7, and 11...\n")
print("DEGREE 1: Linear fit (underfitting)")
plot_polyfit(1)

print("\nDEGREE 7: Good fit (near optimal)")
plot_polyfit(7)

print("\nDEGREE 11: Overfitting to noise")
plot_polyfit(11)
```

## Exercise 5: Cross-Validation to Find Optimal Degree

```python
# SOLUTION: Define RMSE function
def rmse(y_true: np.ndarray, y_pred: np.ndarray) -> float:
    """
    Calculate Root Mean Squared Error.
    
    Parameters:
        y_true: True values
        y_pred: Predicted values
    
    Returns:
        float: RMSE value
    """
    mse = np.mean((y_true - y_pred) ** 2)
    return np.sqrt(mse)

# Test RMSE function
y_test_sample = np.array([1.0, 2.0, 3.0])
y_pred_sample = np.array([1.1, 2.2, 2.9])
rmse_sample = rmse(y_test_sample, y_pred_sample)
print(f"Test RMSE calculation: {rmse_sample:.6f}")
```

```python
# SOLUTION: Compute RMSE for each polynomial degree
print("Computing RMSE for polynomial degrees 1 to 11...\n")

degrees = range(1, 12)
rows = []  # Each row: (degree, rmse_train, rmse_test)

for degree in degrees:
    # Fit polynomial on training data
    poly = polynomial_fit(degree)
    
    # Evaluate on train set
    y_train_pred = poly(x_train)
    rmse_train = rmse(y_train, y_train_pred)
    
    # Evaluate on test set
    y_test_pred = poly(x_test)
    rmse_test = rmse(y_test, y_test_pred)
    
    # Store results
    rows.append((degree, rmse_train, rmse_test))

print("First 3 rows (degrees 1-3):")
for i in range(3):
    print(f"  Degree {rows[i][0]}: Train RMSE = {rows[i][1]:.6f}, Test RMSE = {rows[i][2]:.6f}")

print("\n...")
print("\nLast 3 rows (degrees 9-11):")
for i in range(-3, 0):
    print(f"  Degree {rows[i][0]}: Train RMSE = {rows[i][1]:.6f}, Test RMSE = {rows[i][2]:.6f}")
```

```python
# Create comprehensive results table
results_df = pd.DataFrame(rows, columns=['Degree', 'Train RMSE', 'Test RMSE'])

print("\nCOMPLETE RMSE ANALYSIS (All Degrees)")
print("="*70)
print(results_df.to_string(index=False))

# Find optimal degree
optimal_degree = results_df.loc[results_df['Test RMSE'].idxmin(), 'Degree']
optimal_rmse = results_df['Test RMSE'].min()

print(f"\n" + "="*70)
print(f"OPTIMAL POLYNOMIAL DEGREE: {int(optimal_degree)}")
print(f"Minimum Test RMSE: {optimal_rmse:.6f}")
print(f"This matches the true model (y = -x²) with degree 2!")
print("="*70)
```

```python
# PREFILLED: Plot RMSE curves
degs = [r[0] for r in rows]
rmse_tr = [r[1] for r in rows]
rmse_te = [r[2] for r in rows]

plt.figure(figsize=(10, 6))
plt.plot(degs, rmse_tr, marker='o', linewidth=2.5, markersize=8, 
         label='Train RMSE', color='#4ECDC4', markeredgecolor='black', markeredgewidth=1.5)
plt.plot(degs, rmse_te, marker='^', linewidth=2.5, markersize=8, 
         label='Test RMSE', color='#FF6B6B', markeredgecolor='black', markeredgewidth=1.5)

# Highlight optimal degree
optimal_idx = int(optimal_degree) - 1
plt.scatter([degs[optimal_idx]], [rmse_te[optimal_idx]], s=300, color='gold', 
           marker='*', edgecolor='red', linewidth=2, zorder=5, label=f'Optimal (Degree {int(optimal_degree)})')

plt.yscale('log')
plt.xlabel('Polynomial Degree', fontsize=12, fontweight='bold')
plt.ylabel('RMSE (log scale)', fontsize=12, fontweight='bold')
plt.title('RMSE vs Polynomial Degree - Cross-Validation Analysis', fontsize=13, fontweight='bold')
plt.grid(True, alpha=0.3, which='both')
plt.legend(fontsize=11, loc='best')
plt.xticks(degs)
plt.tight_layout()
plt.show()

print("\nRMSE curves plotted successfully!")
```
