```python
# Exercise 1: Understanding Data Visualization

# Why is data visualization important?
# Data visualization helps transform complex data into easy-to-understand visual forms.
# It allows analysts to identify patterns, trends, and outliers that may not be obvious in raw data.
# In short, it helps in better decision-making.

# What is the purpose of a line graph?
# A line graph is used to show changes or trends over a continuous range — usually time.
# It connects individual data points to show progression clearly.
```

### Exercise 2: Creating a Line Plot for Temperature Variation

```python
import matplotlib.pyplot as plt
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
temperatures = [72, 74, 76, 80, 82, 78, 75]


plt.plot(days, temperatures, marker='o', color='teal')


plt.xlabel("Day")
plt.ylabel("Temperature (°F)")
plt.title("Temperature Variation Over a Week")


plt.show()
```

## Exercise 3: Visualizing Monthly Sales with a Bar Chart


```python
months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [5000, 5500, 6200, 7000, 7500]

# Creating bar chart
plt.bar(months, sales, color='orange')

# Adding labels and title
plt.xlabel("Month")
plt.ylabel("Sales Amount ($)")
plt.title("Monthly Sales Data")

plt.show()
```

### Exercise 4: Visualizing the Distribution of CGPA

```python
import seaborn as sns
import pandas as pd

df = pd.read_csv("student_mental_health.csv")
df.head()
```

```python
# Histogram for CGPA distribution
sns.histplot(df["CGPA"], bins=10, color="skyblue", kde=True)

plt.title("Distribution of Students' CGPA")
plt.xlabel("CGPA")
plt.ylabel("Count")

plt.show()
```

### Exercise 5: Comparing Anxiety Levels Across Different Genders


```python
# Bar plot to compare anxiety across genders
sns.countplot(data=df, x="Choose your gender", hue="Do you have Anxiety?", palette="Set2")

plt.title("Comparison of Anxiety Levels by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Students")

plt.show()
```

### Exercise 6: Exploring the Relationship Between Age and Panic Attacks

```python
# Converting panic attack responses to numeric values
df["Panic_Numeric"] = df["Do you have Panic Attacks?"].map({"Yes": 1, "No": 0})

# Creating scatter plot
sns.scatterplot(data=df, x="Age", y="Panic_Numeric", hue="Panic_Numeric", palette="coolwarm", alpha=0.7)

plt.title("Relationship Between Age and Panic Attacks")
plt.xlabel("Age")
plt.ylabel("Panic Attacks (1=Yes, 0=No)")

plt.show()
```
