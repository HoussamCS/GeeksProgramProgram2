```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler
from sklearn.decomposition import PCA
```

```python
df = pd.read_csv("datascience_salaries.csv")

print("Initial shape:", df.shape)
df.head()
```

```python
print("Missing values per column:\n")
print(df.isnull().sum())
```

```python
scaler = MinMaxScaler()
df["salary_normalized"] = scaler.fit_transform(df[["salary"]])

# Compare before/after
df[["salary", "salary_normalized"]].head(10)
```

## PCA 

```python
# Select only numeric features
numeric_df = df.select_dtypes(include=[np.number])

# PCA to 2 principal components
pca = PCA(n_components=2)
pca_result = pca.fit_transform(numeric_df.fillna(0))

df["PCA1"] = pca_result[:, 0]
df["PCA2"] = pca_result[:, 1]

print("Explained variance ratio:", pca.explained_variance_ratio_)
df[["PCA1", "PCA2"]].head()
```

## Group by experience level 

```python
salary_group = df.groupby("experience_level")["salary"].agg(["mean", "median"]).reset_index()
salary_group
```

```python
print(" Insights")
print("1 Salary normalization rescales all salaries between 0 and 1.")
print("2 PCA reduced numeric data to 2 main dimensions while retaining most variance.")
print("3 Grouping by experience level shows salary evolution across job seniority.")

# Save processed datasets
df.to_csv("processed_datascience_salaries.csv", index=False)
salary_group.to_csv("salary_by_experience.csv", index=False)
print("\n✅ Files saved successfully.")
```
