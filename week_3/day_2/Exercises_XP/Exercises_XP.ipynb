#  Exercise 5 : Export a DataFrame to Excel and JSON

```python


import pandas as pd

data = {
    "Name": ["Alice", "Bob", "Charlie"],
    "Age": [25, 30, 35],
    "Country": ["USA", "UK", "Canada"]
}

df = pd.DataFrame(data)

# Exporting to Excel
excel_path = "simple_data.xlsx"
df.to_excel(excel_path, index=False)
print(f"✅ Excel file saved as: {excel_path}")

# Exporting to JSON
json_path = "simple_data.json"
df.to_json(json_path, orient="records", indent=4)
print(f"✅ JSON file saved as: {json_path}")

df
```
