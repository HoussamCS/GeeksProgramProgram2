# Exercise 1: Generic Exploratory Data Analysis Template (Data-driven decision making)


```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)

# 1. Basic Overview
print("Shape:", df.shape)
display(df.head())
display(df.info())
display(df.describe())

# 2. Missing Values
print("\nMissing Values per Column:")
print(df.isnull().sum())

# 3. Cleaning (example imputation)
df['Age'].fillna(df['Age'].median(), inplace=True)
df['Embarked'].fillna(df['Embarked'].mode()[0], inplace=True)

# 4. Univariate Analysis
plt.figure(figsize=(8,5))
sns.histplot(df['Age'], kde=True, color='skyblue')
plt.title("Age Distribution")
plt.show()

# 5. Bivariate Analysis
plt.figure(figsize=(6,4))
sns.barplot(data=df, x='Sex', y='Survived')
plt.title("Survival Rate by Gender")
plt.show()

# 6. Correlation Heatmap
plt.figure(figsize=(8,5))
sns.heatmap(df.corr(), annot=True, cmap='coolwarm')
plt.title("Correlation Matrix")
plt.show()

# 7. Insights 
print("✅ Key Insights:")
print("- Women had a higher survival rate than men.")
print("- Younger passengers had slightly higher survival rates.")
print("- Passengers in 1st class were more likely to survive.")
```

# Exercise 2: Titanic Dataset Exploratory Data Analysis

```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
titanic = pd.read_csv(url)

# Inspect Data
display(titanic.head())
titanic.info()
titanic.describe()

#  Missing Values
print("\nMissing Values:")
print(titanic.isnull().sum())

#  Cleaning
titanic['Age'].fillna(titanic['Age'].median(), inplace=True)
titanic['Embarked'].fillna(titanic['Embarked'].mode()[0], inplace=True)
titanic.drop(columns=['Cabin'], inplace=True)

#  Encode categorical data
titanic['Sex'] = titanic['Sex'].map({'male': 0, 'female': 1})
titanic['Embarked'] = titanic['Embarked'].map({'S': 0, 'C': 1, 'Q': 2})

#  Univariate Analysis
plt.figure(figsize=(6,4))
sns.histplot(titanic['Age'], bins=30, kde=True)
plt.title("Age Distribution")
plt.show()

#  Survival by Gender
plt.figure(figsize=(5,4))
sns.barplot(x='Sex', y='Survived', data=titanic)
plt.title("Survival Rate by Gender (0=Male,1=Female)")
plt.show()

# Survival by Passenger Class
plt.figure(figsize=(5,4))
sns.barplot(x='Pclass', y='Survived', data=titanic)
plt.title("Survival Rate by Passenger Class")
plt.show()

# Correlation Heatmap
plt.figure(figsize=(8,5))
sns.heatmap(titanic.corr(), annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()

#  Key Insights
print("✅ Insights:")
print("- Females had a much higher survival rate.")
print("- 1st class passengers were most likely to survive.")
print("- Younger passengers had a higher chance of survival.")
```

# Exercise 3 – EDA on the Iris Dataset

```python

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv"
iris = pd.read_csv(url)

# Data Overview
display(iris.head())
iris.info()
iris.describe()

# Unique Species
print("Unique Species:", iris['species'].unique())
print("\nDistribution of Species:")
print(iris['species'].value_counts())

# Feature Distributions
iris.hist(figsize=(10,6), color='skyblue', edgecolor='black')
plt.suptitle("Feature Distributions", fontsize=14)
plt.show()

# Pairplot (all relationships)
sns.pairplot(iris, hue='species')
plt.show()

# Boxplots by Species
plt.figure(figsize=(10,6))
sns.boxplot(data=iris, orient="h")
plt.title("Feature Distribution Comparison")
plt.show()

# Correlation Heatmap
plt.figure(figsize=(6,4))
sns.heatmap(iris.corr(), annot=True, cmap='coolwarm')
plt.title("Correlation Matrix")
plt.show()

# Insights
print("✅ Insights:")
print("- Setosa species have smaller petals and sepals.")
print("- Petal length and width are highly correlated.")
print("- Versicolor and Virginica show overlap but differ in petal size.")
```
