```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## Loading dataset

```python
sleep_df = pd.read_csv("TimeAmericansSpendSleeping.csv")

sleep_df.head(20)
```

```python
sleep_df.info()
sleep_df.describe(include='all')
```

## Handling missing values

```python
sleep_df.isnull().sum()
```

## Quantitative vs Qualitative Data

```python
for col in sleep_df.columns:
    dtype = sleep_df[col].dtype
    if pd.api.types.is_numeric_dtype(sleep_df[col]):
        print(f"{col}: Quantitative (numeric)")
    else:
        print(f"{col}: Qualitative (categorical)")
```

## Statistical Summary

```python
sleep_df.describe()
```

## Visualization (Average Sleep by Category)

```python
plt.figure(figsize=(8,5))
sns.barplot(data=sleep_df, x='Age Group', y='Avg hrs per day sleeping', hue='Sex')
plt.title("Average Sleep Hours by Age Group and Gender")
plt.xlabel("Age Group")
plt.ylabel("Avg hrs per day sleeping")
plt.legend(title='Gender', loc='best')
plt.show()
```

### Observations
- Younger groups tend to sleep more hours than older groups.  
- There may be slight differences in sleep time between genders.  
- The dataset provides clear insights into how sleep duration varies across demographics in the U.S.
