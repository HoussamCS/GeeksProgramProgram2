### Exercice 1

```python
def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True


num = 7
if is_prime(num):
    print(f"{num} is a prime number ")
else:
    print(f"{num} is not a prime number ")
```

### Exercice 2

```python
divisible_by = 3
numbers = [n for n in range(1, 21) if n % divisible_by == 0]

print(f"Numbers between 1 and 20 divisible by {divisible_by}:")
print(numbers)
```
