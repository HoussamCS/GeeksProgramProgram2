# Daily Challenge: Preprocess & Fine-tune Transformer-based Models
This notebook covers preprocessing, tokenization, input formatting, cross-validation, and fine-tuning BERT/XLM-RoBERTa for text classification.

## 1. Understanding BERT and XLM-RoBERTa
- **BERT** is a bidirectional transformer encoder pretrained on large corpora using Masked Language Modeling and Next Sentence Prediction. It excels at understanding context and relationships in text.
- **XLM-RoBERTa** is a multilingual variant of RoBERTa, trained on 100+ languages with improved data and dynamic masking. It is robust for cross-lingual tasks.
- Both models use tokenization to split text into subword units, map them to IDs, and add special tokens for sentence boundaries.

```python
from transformers import BertTokenizer, XLMRobertaTokenizer
bert_tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')
xlmr_tokenizer = XLMRobertaTokenizer.from_pretrained('xlm-roberta-base')
```

```python
# 2. Tokenizing Text
sentence = 'Transformers are powerful models for NLP.'
bert_enc = bert_tokenizer.encode_plus(sentence, max_length=16, padding='max_length', truncation=True, return_tensors='pt')
xlmr_enc = xlmr_tokenizer.encode_plus(sentence, max_length=16, padding='max_length', truncation=True, return_tensors='pt')
print('BERT input_ids:', bert_enc['input_ids'])
print('BERT attention_mask:', bert_enc['attention_mask'])
print('Decoded:', bert_tokenizer.decode(bert_enc['input_ids'][0]))
print('XLM-R input_ids:', xlmr_enc['input_ids'])
print('XLM-R attention_mask:', xlmr_enc['attention_mask'])
print('Decoded:', xlmr_tokenizer.decode(xlmr_enc['input_ids'][0]))
```

```python
# 3. Preparing Input Data for the Model
print('BERT special tokens:', bert_tokenizer.special_tokens_map)
print('BERT vocab size:', bert_tokenizer.vocab_size)
print('XLM-R special tokens:', xlmr_tokenizer.special_tokens_map)
print('XLM-R vocab size:', xlmr_tokenizer.vocab_size)
```

```python
# 4. Loading and Exploring the Dataset
import pandas as pd
train_df = pd.read_csv('train.csv')
test_df = pd.read_csv('test.csv')
print(train_df.head())
print(train_df.shape)
print(test_df.head())
print(test_df.shape)
```

```python
# 5. Creating Cross-Validation Folds
from sklearn.model_selection import StratifiedKFold
kf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
folds = []
for train_idx, val_idx in kf.split(train_df, train_df['label']):
    train_fold = train_df.iloc[train_idx]
    val_fold = train_df.iloc[val_idx]
    folds.append((train_fold, val_fold))
print(f'Created {len(folds)} folds. Each fold maintains label distribution.')
```
