## Exercise 1: Traditional vs. Modern NLP: Comparative Analysis
| Aspect                | Traditional NLP                | Modern NLP                        |
|-----------------------|-------------------------------|-----------------------------------|
| Feature Engineering   | Manual, rule-based            | Automatic, learned from data      |
| Word Representations  | Static (e.g., one-hot, word2vec) | Contextual (e.g., BERT, GPT)     |
| Model Architectures   | Shallow (e.g., SVM, NB)       | Deep (e.g., Transformers)         |
| Training Methodology  | Task-specific, supervised      | Pre-training + fine-tuning        |
| Key Examples          | Naïve Bayes, SVM, LSTM        | BERT, GPT, T5                     |
| Advantages            | Simpler, interpretable, fast   | Scalable, robust, transfer learning|
| Disadvantages         | Limited context, manual effort | Data-hungry, less interpretable   |

**Discussion:** Modern NLP leverages representation learning and deep architectures, enabling models to learn features automatically and capture context. Hardware parallelism (GPUs/TPUs) allows training on massive datasets, improving scalability. Pretraining and transfer learning reduce the need for labeled data and enable rapid adaptation to new tasks. As a result, NLP applications are more efficient, accurate, and scalable than ever before.

## Exercise 2: LLM Architecture and Application Scenarios
**BERT:** Bidirectional encoder-only; uses Masked Language Modeling (MLM) and (historically) Next Sentence Prediction (NSP).
Application: Text classification, Named Entity Recognition, Question Answering.
Why: Bidirectional context enables understanding of relationships in text, making it ideal for tasks needing deep comprehension.

**GPT:** Decoder-only; autoregressive causal LM.
Application: Chatbots, text generation, code completion.
Why: Unidirectional, left-to-right generation is suited for producing coherent, creative text outputs.

**T5:** Encoder-decoder; text-to-text framework.
Application: Summarization, translation, multi-task learning.
Why: Flexible input/output format and encoder-decoder structure allow handling diverse NLP tasks.

## Exercise 3: Benefits and Ethics of Pre-training
**Benefits:**
- Improved generalization: Models learn broad language patterns, making them effective on unseen data.
- Less labeled data: Pretraining reduces the need for large labeled datasets for downstream tasks.
- Faster fine-tuning: Pretrained models adapt quickly to new tasks, saving time and resources.
- Transfer learning: Knowledge from pretraining transfers to many applications, boosting performance.
- Robustness: Exposure to diverse data makes models resilient to noise and domain shifts.
**Ethical concerns:** Bias, misinformation, misuse, privacy risks.
**Mitigations:** Careful data curation, differential privacy, safety filters, RLHF (Reinforcement Learning from Human Feedback), regular audits.

## Exercise 4: Transformer Architecture Deep Dive
**Self-Attention & Multi-Head Attention:**
- Self-attention computes Q/K/V for each token, scores relevance via softmax, and mixes values to create context-aware representations.
- Multi-head attention projects Q/K/V into multiple subspaces, allowing the model to capture diverse relationships (e.g., subject-verb, object-modifier) simultaneously.
- Example: In 'The quick brown fox jumps over the lazy dog', one head may focus on 'fox-jumps', another on 'dog-lazy', capturing different dependencies.
**Pre-training Objectives:**
- MLM masks random tokens for prediction (bidirectional context); CLM predicts next token (unidirectional).
- MLM is best for understanding/filling gaps (e.g., classification); CLM for generation (e.g., story writing).
- Early BERT used NSP to model sentence relationships, but modern models avoid it due to limited benefit and possible noise.
**Transformer Model Selection:**
- Sentiment analysis: Encoder-only (BERT) for deep understanding.
- Chatbot: Decoder-only (GPT) for creative generation.
- Translation: Encoder-decoder (T5) for mapping input to output sequences.
**Positional Encoding:**
- Adds position info to token embeddings, enabling transformers to distinguish word order.
- Without it, 'dog bites man' and 'man bites dog' would be indistinguishable.

## Exercise 6: Softmax Temperature - The Randomness Regulator
**Temperature Scenarios:**
- T=0.2: Output is highly deterministic, repetitive, and conservative.
- T=1.5: Output is creative, diverse, and sometimes incoherent.
- T=1.0: Balanced creativity and coherence.
**Application Design:**
- Bedtime stories: Use higher temperature (e.g., 1.2) for creativity, lower (e.g., 0.7) for coherence.
- Financial summaries: Use low temperature (e.g., 0.3) for accuracy and reliability.
**Temperature & Bias:**
- Lower temperature amplifies most likely (and possibly biased) outputs; higher temperature increases diversity, potentially surfacing less common perspectives.
- Example: With low temperature, a model may always describe a CEO as male; higher temperature may yield more varied outputs.

```python
# (Optional) Quick Generation Demo Across Temperatures
from transformers import AutoTokenizer, AutoModelForCausalLM, pipeline
import torch
model_name = 'gpt2'
prompt = 'Artificial intelligence will'
pipe = pipeline('text-generation', model=AutoModelForCausalLM.from_pretrained(model_name), tokenizer=AutoTokenizer.from_pretrained(model_name), device=0 if torch.cuda.is_available() else -1)
for T in [0.2, 1.0, 1.5]:
    out = pipe(prompt, max_new_tokens=40, temperature=T, do_sample=True)
    print('--- Temperature:', T)
    print(out[0]['generated_text'])
```
