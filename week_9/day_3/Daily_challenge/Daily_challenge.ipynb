# Evaluating Large Language Models
Comprehensive exercises on LLM evaluation: BLEU, ROUGE, Perplexity, adversarial testing, and human evaluation.

## 1. Understanding LLM Evaluation
- Evaluating LLMs is more complex than traditional software because outputs are probabilistic, context-dependent, and open-ended, making correctness hard to define.
- Key reasons for evaluating LLM safety: prevent harmful outputs, ensure reliability, avoid bias, and maintain ethical standards.
- Adversarial testing exposes weaknesses by presenting tricky or ambiguous prompts, helping improve model robustness.
- Automated metrics (BLEU, ROUGE) focus on content overlap but miss nuances like creativity, context, and factuality; human evaluation captures these but is costly and subjective.

```python
# Applying BLEU and ROUGE Metrics
from nltk.translate.bleu_score import sentence_bleu, SmoothingFunction
import evaluate
reference_bleu = ['Despite the increasing reliance on artificial intelligence in various industries, human oversight remains essential to ensure ethical and effective implementation.']
generated_bleu = 'Although AI is being used more in industries, human supervision is still necessary for ethical and effective application.'
bleu_score = sentence_bleu([r.split() for r in reference_bleu], generated_bleu.split(), smoothing_function=SmoothingFunction().method1)
print(f'BLEU score: {bleu_score:.4f}')
reference_rouge = 'In the face of rapid climate change, global initiatives must focus on reducing carbon emissions and developing sustainable energy sources to mitigate environmental impact.'
generated_rouge = 'To counteract climate change, worldwide efforts should aim to lower carbon emissions and enhance renewable energy development.'
rouge = evaluate.load('rouge')
rouge_score = rouge.compute(predictions=[generated_rouge], references=[reference_rouge])
print('ROUGE score:', rouge_score)
```

### Limitations of BLEU and ROUGE
- Both metrics rely on n-gram overlap, which penalizes creative or contextually correct paraphrasing.
- They do not account for factual accuracy, coherence, or semantic meaning.
- Improvements: Use metrics like BERTScore, METEOR, or combine with human evaluation for richer assessment.

## 3. Perplexity Analysis
- Perplexity measures how well a model predicts a sample. Lower perplexity means higher confidence.
- Model A (0.8) has lower perplexity than Model B (0.4) for 'mitigation', meaning Model A is more confident.
- A perplexity score of 100 indicates poor performance; the model is uncertain. To improve: use more data, better architecture, or regularization.

## 6. Comparative Analysis of Evaluation Methods
- Task: Text Summarization
- BLEU: Measures n-gram overlap, good for translation but less for summarization.
- ROUGE: Measures recall of n-grams, widely used for summarization.
- BERTScore: Uses contextual embeddings for semantic similarity.
- Human Evaluation: Captures fluency, coherence, and factuality.
- Most appropriate: ROUGE and BERTScore for summarization, as they balance overlap and semantic meaning. Human evaluation is best for final assessment.
