# Generative AI - Daily Challenge Answers

## 1) Compare Traditional AI vs Generative AI

**Three scenarios where traditional AI is more effective**
1. **Medical diagnosis from scans**: classification models are easier to validate, explain, and certify for safety.
2. **Fraud detection**: anomaly classifiers on transactions provide stable, auditable decisions.
3. **Autonomous driving perception**: real-time object detection and tracking need high reliability and low latency.

**Three scenarios where generative AI is more effective**
1. **Content drafting**: producing first drafts of articles, emails, and scripts.
2. **Data augmentation**: synthesizing training data when real samples are scarce (e.g., rare diseases).
3. **Creative design**: generating images, music, or UI concepts from prompts.

## 2) Benefits and Risks of Generative AI in Content Creation

**Benefits**
- Faster production of text, images, and audio.
- Lower cost for prototyping ideas.
- Enables personalization at scale (marketing, education).

**Risks**
- **Misinformation**: realistic but false content can spread quickly.
- **Copyright/IP issues**: outputs may closely resemble protected works.
- **Bias and stereotyping**: training data can embed harmful patterns.

**Can it be used for misleading or unethical content? How?**
Yes. It can generate deepfakes, fake news articles, impersonation emails, and synthetic reviews. The realism makes deception harder to detect, especially without provenance and watermarking.

## Conclusion
Generative AI excels at creating new content and accelerating creativity, while traditional AI remains stronger for stable, high-stakes decisions. Responsible use requires verification, transparency, and safeguards against misuse.
