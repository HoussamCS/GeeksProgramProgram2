# Generative AI Exercises - Answers

## Exercise 1: Comparative Analysis of Generative AI and Traditional AI

1. **Automated Medical Diagnosis**
   - **Prefer:** Traditional AI (supervised classification) for diagnosis.
   - **Why:** Prioritizes accuracy, explainability, and safety. Generative models can help with data augmentation, but diagnosis should rely on validated classifiers with strict validation. Generative synthesis adds risk of distribution shift and subtle artifacts.

2. **Legal Document Generation**
   - **Prefer:** Hybrid, but core should be traditional rules + human review; generative AI only as a draft assistant.
   - **Why:** Contracts demand correctness and compliance. LLMs can hallucinate or omit clauses. Risks: legal liability, privacy leaks, ambiguous language, inconsistent definitions.

3. **AI-Generated Scientific Research**
   - **Prefer:** Generative AI for summarization and hypothesis brainstorming, with strict human verification.
   - **Why:** LLMs can reduce reading time and connect concepts, but they can fabricate citations or overstate results. Use retrieval + citations and require expert review.

4. **Financial Market Predictions**
   - **Prefer:** Traditional AI (time-series, regression, tree models). Generative AI can be used for scenario simulation but not primary signal.
   - **Why:** Markets need stable, measurable performance and low-latency inference. Generative models are costly and can be unstable; they add risk of spurious patterns.

5. **Autonomous Vehicle Decision-Making**
   - **Prefer:** Traditional AI + safety-verified control systems; limited generative use for simulation.
   - **Why:** Real-time safety and reliability are critical. Generative models are harder to certify and can be unpredictable. Use generative AI mainly for data augmentation and simulation.

## Exercise 2: Ethical and Security Risks of Generative AI

1. **Deepfake Political Manipulation**
   - **Risks:** Misinformation, voter manipulation, erosion of trust, social unrest.
   - **Mitigations:** Watermarking + provenance (C2PA), rapid takedown policies, media literacy campaigns, detection models in platforms.
   - **Consequences if unregulated:** Election integrity damage and long-term distrust.

2. **Synthetic Identity Fraud**
   - **Risks:** Financial fraud, account takeover, bypass of biometric systems.
   - **Mitigations:** Multi-factor verification, liveness detection, behavioral biometrics, fraud monitoring.
   - **Consequences:** Systemic fraud, loss of consumer confidence.

3. **Generative AI in Cyber Warfare**
   - **Risks:** Deception, escalation, misattribution, policy errors based on fake intel.
   - **Mitigations:** Human verification loops, source validation, cross-agency corroboration, legal frameworks.
   - **Consequences:** Diplomatic crises, conflict escalation.

4. **AI-Generated Malware**
   - **Risks:** Novel evasion techniques, rapid mutation, large-scale attacks.
   - **Mitigations:** Behavior-based detection, sandboxing, AI red-teaming, strict model access controls.
   - **Consequences:** Increased attack success rates and infrastructure damage.

5. **Copyright and IP Theft**
   - **Risks:** IP infringement, economic harm to creators, legal disputes.
   - **Mitigations:** Opt-in training data, licensing, dataset audits, output similarity checks.
   - **Consequences:** Litigation, regulatory penalties, loss of trust.

## Exercise 3: Optimization and Fine-Tuning of Generative AI Models

1. **Prompt Engineering (rewrite prompts)**
   - Futuristic city image: "Create a 4k, dusk-time city skyline with flying vehicles, neon signage, and clean energy infrastructure, in a semi-realistic style, wide-angle view."
   - Poem about the future: "Write a 12-line poem in free verse about human-AI collaboration, with a hopeful tone and one recurring metaphor of light."
   - Classical music song: "Compose a 90-second classical piece in the style of Mozart, with a clear ABA structure, strings and woodwinds only, and a lively allegro tempo."

2. **Bias and Fairness in Training Data**
   - Use diverse sources across political spectrum.
   - Balance topics, regions, and time ranges.
   - Remove duplicated or extreme outliers.
   - Add bias evaluation sets and monitor model outputs.

3. **Fine-Tuning for Medical Research Papers**
   - **Data:** Curated, peer-reviewed papers with strong metadata.
   - **Preprocess:** De-dup, normalize citations, filter retracted studies.
   - **Transfer learning:** Start from a strong base model, apply domain-adaptive pretraining, then supervised fine-tuning.
   - **Evaluation:** Factuality checks, citation grounding, expert review, and error analysis.

4. **Quantitative Metrics**
   - Text: Perplexity, BLEU, ROUGE, BERTScore, factuality/QA-based scoring.
   - Images: FID, IS (Inception Score), precision/recall for generative models.
   - Audio: Fréchet Audio Distance (FAD).

5. **Controlling Creativity and Coherence**
   - **Temperature:** Lower for coherence, higher for diversity.
   - **RLHF / RLAIF:** Reward factuality and consistency.
   - **Attention mechanisms:** Improve long-range coherence in outputs.

## Exercise 4: Trade-offs Between GANs and VAEs

1. **Synthetic Medical Image Generation**
   - **Prefer:** VAEs for interpretability and stable training; GANs if high realism is essential.

2. **AI-Assisted Creative Writing**
   - **Prefer:** VAEs for controllable latent space and smoother interpolation of style.

3. **Anomaly Detection in Financial Transactions**
   - **Prefer:** VAEs, since reconstruction error supports anomaly scoring.

4. **High-Resolution Fashion Designs**
   - **Prefer:** GANs for sharp, realistic images; VAEs for style interpolation.

5. **Autonomous Vehicle Data Augmentation**
   - **Prefer:** GANs for realism, VAEs for coverage of edge cases; often a hybrid approach is best.

## Exercise 5: Latent Space Exploration in VAEs

1. **Visualizing Latent Space**
   - Check class separation via 2D projections (t-SNE, UMAP).
   - Use clustering like K-means or Gaussian Mixture Models.

2. **Interpolation**
   - Encode images to latent vectors z1 and z2, then linearly interpolate: z = (1-t)z1 + t z2.
   - VAEs enforce a continuous latent space; GANs do not guarantee smooth latent structure.

3. **Controlling Variability**
   - Increase KL term -> more regularization, smoother but less sharp outputs.
   - Decrease KL term -> sharper reconstructions but less diverse outputs.
   - Example: Medical imaging where you want controlled variability for safe augmentation.

4. **Disentangling Latents**
   - Use beta-VAE or disentanglement constraints.
   - Adjust one latent dimension tied to hair color while holding others fixed.

5. **PCA vs VAEs**
   - **Linearity:** PCA is linear; VAEs are nonlinear.
   - **Interpretability:** PCA components are explicit; VAE latents are learned and less interpretable.
   - **Reconstruction:** VAEs handle complex data better.
   - **Generative use:** VAEs can sample; PCA cannot generate realistic new data.
