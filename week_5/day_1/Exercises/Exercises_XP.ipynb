##  Exercise 1: Array Creation and Manipulation


```python
array_1d = np.arange(10)

print("Exercise 1 - Creating 1D array from 0 to 9:")
print(array_1d)
print(f"Array shape: {array_1d.shape}")
print(f"Array type: {type(array_1d)}")
```

##  Exercise 2: Type Conversion and Array Operations


```python
original_list = [3.14, 2.17, 0, 1, 2]
float_array = np.array(original_list)
int_array = float_array.astype(int)

print("Exercise 2 - Type conversion:")
print(f"Original list: {original_list}")
print(f"Float array: {float_array}")
print(f"Integer array: {int_array}")
print(f"Original dtype: {float_array.dtype}")
print(f"Converted dtype: {int_array.dtype}")
```

##  Exercise 3: Working with Multi-Dimensional Arrays


```python
# Creating 3x3 array with values from 1 to 9 using arange and reshape
array_3x3 = np.arange(1, 10).reshape(3, 3)

print("Exercise 3 - Creating 3x3 array:")
print(array_3x3)
print(f"Array shape: {array_3x3.shape}")
print(f"Array dimensions: {array_3x3.ndim}")
print(f"Total elements: {array_3x3.size}")
```

##  Exercise 4: Creating Multi-Dimensional Array with Random Numbers


```python
np.random.seed(42)

# Creating 4x5 array with random numbers between 0 and 1
random_array = np.random.rand(4, 5)

print("Exercise 4 - Creating 4x5 random array:")
print(random_array)
print(f"Array shape: {random_array.shape}")
print(f"Min value: {random_array.min():.4f}")
print(f"Max value: {random_array.max():.4f}")
```

##  Exercise 5: Indexing Arrays


```python
array = np.array([[21,22,23,22,22],[20, 21, 22, 23, 24],[21,22,23,22,22]])

# Selecting the second row (index 1)
second_row = array[1]

print("Exercise 5 - Indexing arrays:")
print("Original array:")
print(array)
print(f"\nSecond row (index 1): {second_row}")
print(f"Shape of second row: {second_row.shape}")
print(f"Type of second row: {type(second_row)}")
```

##  Exercise 6: Reversing Elements


```python
# Creating array from 0 to 9 and reversing it
original_array = np.arange(10)
reversed_array = original_array[::-1]

print("Exercise 6 - Reversing elements:")
print(f"Original array: {original_array}")
print(f"Reversed array: {reversed_array}")

# Alternative methods for reversing
reversed_with_flip = np.flip(original_array)
print(f"Using np.flip(): {reversed_with_flip}")
```

##  Exercise 7: Identity Matrix


```python
# Creating 4x4 identity matrix
identity_matrix = np.eye(4)

print("Exercise 7 - Creating 4x4 identity matrix:")
print(identity_matrix)
print(f"Matrix shape: {identity_matrix.shape}")
print(f"Data type: {identity_matrix.dtype}")

# Alternative method using identity function
identity_alt = np.identity(4)
print(f"\nUsing np.identity(): Same result: {np.array_equal(identity_matrix, identity_alt)}")
```

##  Exercise 8: Simple Aggregate Functions


```python
# Creating array from 0 to 9 for sum and average calculation
numbers_array = np.arange(10)

# Calculating sum and average
array_sum = np.sum(numbers_array)
array_average = np.mean(numbers_array)

print("Exercise 8 - Simple aggregate functions:")
print(f"Array: {numbers_array}")
print(f"Sum: {array_sum}, Average: {array_average}")

# Additional statistics
print(f"Median: {np.median(numbers_array)}")
print(f"Standard deviation: {np.std(numbers_array):.2f}")
print(f"Min: {np.min(numbers_array)}, Max: {np.max(numbers_array)}")
```

##  Exercise 9: Create Array and Change its Structure


```python
# Creating array from 1 to 20 and reshaping to 4x5 matrix
linear_array = np.arange(1, 21)
matrix_4x5 = linear_array.reshape(4, 5)

print("Exercise 9 - Creating and reshaping array:")
print(f"Original 1D array: {linear_array}")
print(f"\nReshaped 4x5 matrix:")
print(matrix_4x5)
print(f"Original shape: {linear_array.shape}")
print(f"New shape: {matrix_4x5.shape}")
print(f"Total elements: {matrix_4x5.size}")
```

##  Exercise 10: Conditional Selection of Values


```python
# Creating array from 0 to 9 and extracting odd numbers
source_array = np.arange(10)
odd_numbers = source_array[source_array % 2 == 1]

print("Exercise 10 - Conditional selection:")
print(f"Original array: {source_array}")
print(f"Odd numbers: {odd_numbers}")

# Alternative approach using where function
odd_with_where = np.where(source_array % 2 == 1, source_array, -1)
odd_filtered = odd_with_where[odd_with_where != -1]
print(f"Using np.where: {odd_filtered}")

even_numbers = source_array[source_array % 2 == 0]
greater_than_5 = source_array[source_array > 5]
print(f"Even numbers: {even_numbers}")
print(f"Numbers > 5: {greater_than_5}")
```
