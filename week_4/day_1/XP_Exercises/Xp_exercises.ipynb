## Exercise 1: Understanding Data Visualization

##### Q1: Why is data visualization important in data analysis?
##### ➤ Data visualization helps to transform complex data into visual insights.
##### It allows us to identify patterns, trends, and outliers quickly, making the data easier to interpret and act upon.

##### Q2: What is the purpose of a line graph in data visualization?
##### ➤ A line graph shows how a variable changes over time or across categories.
##### It is ideal for representing continuous data like temperature, sales trends, or stock prices.


### Exercise 2: Creating a Line Plot for Temperature Variation

```python
import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
temperatures = [72, 74, 76, 80, 82, 78, 75]

plt.figure(figsize=(8,5))
plt.plot(days, temperatures, marker='o', color='tomato', linestyle='-', linewidth=2)
plt.title("Temperature Variation Over a Week")
plt.xlabel("Day")
plt.ylabel("Temperature (°F)")
plt.grid(True)
plt.show()
```

#### Exercise 3: Visualizing Monthly Sales with a Bar Chart

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [5000, 5500, 6200, 7000, 7500]

plt.figure(figsize=(8,5))
plt.bar(months, sales, color="skyblue", edgecolor="black")
plt.title("Monthly Sales Data")
plt.xlabel("Month")
plt.ylabel("Sales Amount ($)")
plt.show()
```

### Exercise 4: Data Visualization with sales dataset

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("sales_data.csv")

# Total quantity of products sold
total_quantity = df['quantity'].sum()

#  Category with highest revenue
df['revenue'] = df['quantity'] * df['price']  # Assuming dataset has 'price'
highest_revenue_category = df.groupby('category')['revenue'].sum().idxmax()
highest_revenue_amount = df.groupby('category')['revenue'].sum().max()

# Average revenue per sale
average_revenue = df['revenue'].mean()

print(" Total Quantity Sold:", total_quantity)
print(" Highest Revenue Category:", highest_revenue_category, "-", highest_revenue_amount)
print(" Average Revenue per Sale:", round(average_revenue, 2))

# Bar chart of revenue per quarter
df['date'] = pd.to_datetime(df['date'])
df['quarter'] = df['date'].dt.to_period('Q')
quarterly_revenue = df.groupby('quarter')['revenue'].sum()

plt.figure(figsize=(8,5))
plt.bar(quarterly_revenue.index.astype(str), quarterly_revenue.values, color='seagreen')
plt.title("Total Revenue by Quarter")
plt.xlabel("Quarter")
plt.ylabel("Revenue ($)")
plt.show()
```

#### Exercise 5: Data Visualization Using MatPlotLib

```python

import matplotlib.pyplot as plt
import numpy as np

# Line plot for y = x^2
x = np.arange(-10, 11, 1)
y = x**2

plt.figure(figsize=(7,5))
plt.plot(x, y, color='orange', marker='o')
plt.title("y = x² Function")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.show()
```

```python
# 2️ Bar chart for product sales

products = ['A', 'B', 'C', 'D']
sales = [15, 30, 45, 20]

plt.figure(figsize=(7,5))
plt.bar(products, sales, color=['teal', 'orange', 'salmon', 'purple'])
plt.title("Weekly Product Sales")
plt.xlabel("Product")
plt.ylabel("Sales")
plt.show()
```

```python
# 3️ Pie chart for favorite fruits

fruits = ['Apples', 'Bananas', 'Cherries', 'Dates']
percentages = [40, 30, 20, 10]
colors = ['red', 'yellow', 'pink', 'brown']

plt.figure(figsize=(7,7))
plt.pie(percentages, labels=fruits, autopct='%1.1f%%', colors=colors, startangle=140)
plt.title("Favorite Fruits Distribution")
plt.legend(fruits, title="Fruits", loc="best")
plt.show()
```
