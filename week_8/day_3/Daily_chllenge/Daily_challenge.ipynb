# Build a Retrieval-Augmented Generation (RAG) System
This notebook demonstrates how to use Hugging Face Transformers to load a pre-trained BERT model, tokenize input, and obtain contextual embeddings for downstream tasks such as retrieval or classification.

```python
# Install Hugging Face Transformers
!pip install transformers
```

```python
from transformers import AutoTokenizer, AutoModel
tokenizer = AutoTokenizer.from_pretrained('bert-base-uncased')
model = AutoModel.from_pretrained('bert-base-uncased')
```

```python
# Tokenize input and get model output
inputs = tokenizer('Hello Hugging Face!', return_tensors='pt')
outputs = model(**inputs)
embeddings = outputs.last_hidden_state
print('Embeddings shape:', embeddings.shape)
# Each token in the input has a corresponding embedding vector
```

You can use these embeddings for tasks such as classification, semantic search, or as part of a RAG pipeline to retrieve relevant documents and generate context-aware responses.
