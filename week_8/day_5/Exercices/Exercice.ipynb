# Mini-Project: Document-Based Question Answering System (LlamaIndex + HuggingFace)
This notebook builds a retrieval-based QA system using llama_index and HuggingFace models on BBC text documents.

```python
# Step 2 – Import Libraries
from llama_index.core import SimpleDirectoryReader, VectorStoreIndex, Settings
from llama_index.llms.huggingface import HuggingFaceLLM
from llama_index.embeddings.huggingface import HuggingFaceEmbedding
```

```python
# Step 3 – Load Documents
documents = SimpleDirectoryReader('papers').load_data()
```

```python
# Step 4 – Configure the Language Model (LLM)
llm = HuggingFaceLLM(
    model_name='TinyLlama/TinyLlama-1.1B-Chat-v1.0',
    tokenizer_name='TinyLlama/TinyLlama-1.1B-Chat-v1.0',
    context_window=2048,
    max_new_tokens=256,
    device_map='auto'
)
```

```python
# Step 5 – Set the Embedding Model
embed_model = HuggingFaceEmbedding(
    model_name='sentence-transformers/all-MiniLM-L6-v2'
)
```

```python
# Step 6 – Apply Global Settings
Settings.llm = llm
Settings.embed_model = embed_model
```

```python
# Step 7 – Create the Index
index = VectorStoreIndex.from_documents(documents)
```

```python
# Step 8 – Save the Index
index.storage_context.persist(persist_dir='./index_store')
```

```python
# Step 9 – Ask Questions!
query_engine = index.as_query_engine()
response1 = query_engine.query('What are the main points in the article?')
print('Main points:', response1)
response2 = query_engine.query('Summarize key events described.')
print('Key events:', response2)
response3 = query_engine.query('Give a summary of the document in 5 bullet points.')
print('5 bullet points:', response3)
```
