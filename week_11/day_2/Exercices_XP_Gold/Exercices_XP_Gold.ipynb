## Part 3: Load Model and Generate Text

### Step 3.1: Initialize Llama Model
We'll use llama-cpp-python bindings to load the GGUF model with GPU acceleration.

```python
# Step 3.1: Load the GGUF model using llama-cpp-python
from llama_cpp import Llama

# Initialize the model
# n_gpu_layers: number of layers to offload to GPU (adjust based on VRAM)
# n_threads: CPU threads for inference
# verbose: print loading information

print("Loading Gemma 3 1B model...")

llm = Llama(
    model_path="gemma-3-1b-it-Q8_0.gguf",
    n_gpu_layers=32,  # Offload layers to GPU (adjust for your VRAM)
    n_threads=4,      # CPU threads
    verbose=False
)

print("✅ Model loaded successfully!")
print(f"Model type: Gemma 3 1B")
print(f"Context length: {llm.n_ctx()} tokens")
```

```python
# Step 3.2: Generate text with a natural language prompt
print("=" * 60)
print("Test 1: Generate Text - Explain Solar System")
print("=" * 60)

prompt = "Explain how the solar system formed in 3-4 sentences."

output = llm(
    prompt=prompt,
    max_tokens=150,
    temperature=0.7,
    top_p=0.95
)

print("Prompt:", prompt)
print("\nGenerated Text:")
print(output["choices"][0]["text"])
print("\n")
```

```python
# Step 3.3: Generate Python code with streaming inference
print("=" * 60)
print("Test 2: Generate Python Code - Prime Number Checker")
print("=" * 60)

code_prompt = "Write a Python function that checks if a number is prime."

print("Prompt:", code_prompt)
print("\nGenerated Code:")
print("-" * 40)

# Use streaming to output tokens as they're generated
output_stream = llm(
    prompt=code_prompt,
    max_tokens=256,
    temperature=0.5,
    stream=True
)

# Iterate through tokens and print in real-time
for output in output_stream:
    token = output["choices"][0]["text"]
    print(token, end="", flush=True)

print("\n" + "-" * 40)
print("\n✅ Code generation complete!")
```
