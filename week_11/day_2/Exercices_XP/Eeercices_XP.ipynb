# Exercise 1: Open Source Levels Reflection
Identify definitions and key characteristics of Fully Open, Weights Released, and Architecture Only LLMs.

```python
# EXERCISE 1: capture definitions and key characteristics
open_source_levels = {
    "Fully Open": {
        "definition": "Model code, weights, training data, and procedures are all publicly available and modifiable.",
        "what_is_open": ["Model architecture source code", "Pretrained weights and checkpoints"],
        "what_you_can_do": ["Inspect and modify the entire model", "Retrain from scratch or fine-tune on custom data"],
        "what_you_cannot_do": ["Not applicable—you have full control"],
    },
    "Weights Released": {
        "definition": "Pretrained weights are publicly available, but source code and training procedures may be restricted.",
        "what_is_open": ["Pretrained model weights", "Model architecture description (partial or full)"],
        "what_you_can_do": ["Fine-tune on custom data", "Use for inference immediately"],
        "what_you_cannot_do": ["Retrain from scratch without weights", "Modify core architecture without rebuilding"],
    },
    "Architecture Only": {
        "definition": "Only the model design and structure are published; weights and training data remain proprietary.",
        "what_is_open": ["Model architecture design specs", "High-level implementation guidelines"],
        "what_you_can_do": ["Understand the model design", "Reimplement the architecture from scratch"],
        "what_you_cannot_do": ["Use pretrained weights directly", "Quickly deploy without extensive retraining"],
    },
}
open_source_levels
```

```python
# EXERCISE 1: fill the comparison table
comparison_table = """| Openness level | What's open? | Impact on retraining/modifying |
| --- | --- | --- |
| Fully Open | Code, weights, training data, procedures | Full control: inspect, modify, retrain from scratch or fine-tune freely |
| Weights Released | Pretrained weights and model architecture (partial) | Can fine-tune rapidly but cannot deeply modify architecture without rebuilding |
| Architecture Only | Model design and specifications only | Must retrain from scratch; time-consuming but allows custom optimization |"""
print(comparison_table)
```

```python
# EXERCISE 1: 3–5 sentence comparative paragraph
comparative_paragraph = """Open-source LLMs exist on a spectrum of transparency and control. Fully Open models release all code, weights, and training procedures, enabling researchers and organizations to inspect internals, modify behavior, and retrain on custom data—essential for high-security and compliance-heavy domains like healthcare. Weights Released models provide immediately usable pretrained parameters for fine-tuning but restrict access to implementation details, offering a practical middle ground for rapid deployment with domain adaptation. Architecture Only models share design specifications but require organizations to retrain from scratch, demanding significant computational resources but enabling complete customization. For healthcare applications requiring clinical data retraining, Fully Open or Weights Released levels are essential; Fully Open is preferred for regulated environments requiring auditability and compliance verification."""

print(comparative_paragraph)
```
