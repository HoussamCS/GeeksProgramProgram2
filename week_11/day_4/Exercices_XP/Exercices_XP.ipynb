```python
# Different chunking for different content types

CHUNKING_CONFIG = {
    'lecture_slides': {
        'method': 'by_slide',  # Each slide = one chunk
        'overlap': 0,  # Slides are discrete units
        'metadata': ['week', 'day', 'topic', 'slide_number']
    },
    
    'lecture_notes': {
        'method': 'semantic',  # Split by topics/sections
        'chunk_size': 800,  # tokens
        'overlap': 100,  # maintain context
        'metadata': ['week', 'day', 'section_title']
    },
    
    'code_examples': {
        'method': 'by_function',  # Each function/class = chunk
        'include_context': True,  # Include file-level docstrings
        'metadata': ['language', 'topic', 'difficulty', 'week']
    },
    
    'exercises': {
        'method': 'by_exercise',  # Each exercise = one chunk
        'include': ['description', 'hints', 'solution'],
        'metadata': ['exercise_id', 'week', 'difficulty', 'topics']
    },
    
    'qna_pairs': {
        'method': 'by_qa_pair',  # Question + Answer = chunk
        'chunk_size': 500,
        'metadata': ['date', 'topic', 'student_id_hashed']
    },
    
    'documentation': {
        'method': 'semantic',
        'chunk_size': 600,
        'overlap': 150,  # Higher overlap for reference docs
        'metadata': ['library', 'version', 'category']
    }
}
```

#### **Challenge 2: Multi-Step Reasoning Questions**

**Problem:**
- "How do I read a CSV, filter rows where age > 30, and calculate average salary?" requires multiple concepts
- Single retrieval may miss some steps
- Context window limitations

**Solutions:**
- Implement multi-hop retrieval: initial retrieval → extract key concepts → retrieve again
- Query decomposition: break complex questions into sub-questions
- Increase k (retrieve more documents) for complex queries
- Use GPT-4 for complex multi-step questions

```python
def handle_complex_query(query):
    # Decompose into sub-questions
    sub_questions = llm.decompose(query)
    
    # Retrieve for each
    all_context = []
    for sub_q in sub_questions:
        context = retrieve(sub_q, k=2)
        all_context.extend(context)
    
    # Synthesize final answer
    return llm.generate(query, all_context)
```
