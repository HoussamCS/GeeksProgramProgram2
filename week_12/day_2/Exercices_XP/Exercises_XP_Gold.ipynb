````python
def run_simulation(num_rounds: int = 3, num_penguins: int = 4):
    """Run the multi-agent simulation."""
    print("\n" + "="*60)
    print("ANTARCTIC AGENTS: Multi-Agent Simulation")
    print("="*60)
    
    # Initialize agents
    scientist = ScientistAgent(initial_food_supply=20, refresh_interval=5)
    penguins = [PenguinAgent(f"Penguin_{i}") for i in range(1, num_penguins + 1)]
    
    print(f"\nInitialized: 1 Scientist Agent, {num_penguins} Penguin Agents")
    print(f"Scientist starting resources: {scientist.food_supply} food, tool available: {scientist.tool_available}")
    print(f"Each penguin starts with: 5 food, no tool\n")

    # Run simulation rounds
    for round_idx in range(num_rounds):
        print("\n" + "-"*60)
        print(f"ROUND {round_idx + 1} / {num_rounds}")
        print("-"*60)

        # Phase 1: Penguins take actions
        print("\n[PHASE 1: Penguin Actions]")
        penguin_actions = {}
        for penguin in penguins:
            action = penguin.take_action()
            penguin_actions[penguin.name] = action
            print(f"  {penguin.name}: {action['details']}")
            print(f"    Action={action['action']}, Method={action['method']}")

        # Phase 2: Process actions and find food
        print("\n[PHASE 2: Action Resolution]")
        for penguin in penguins:
            act = penguin_actions[penguin.name].get("action")
            if act == "find_food":
                method = penguin_actions[penguin.name].get("method", "foraging")
                try:
                    food_found = find_food(penguin.name, method)
                    penguin.food += food_found
                except Exception as e:
                    print(f"  Error: {e}")
            elif act == "rest":
                print(f"  {penguin.name} is resting...")
            elif act == "request_food":
                print(f"  {penguin.name} is requesting food from scientist...")

        # Phase 3: Scientist responds
        print("\n[PHASE 3: Scientist Response]")
        for penguin in penguins:
            scientist.respond_to_action(penguin, penguin_actions[penguin.name])

        # Print round summary
        print("\n[ROUND SUMMARY]")
        print(f"  Scientist - Food: {scientist.food_supply}, Tool: {scientist.tool_available}")
        for penguin in penguins:
            hist = check_history(penguin.name)
            print(f"  {penguin.name} - Food: {penguin.food}, Tool: {penguin.has_tool}, Received: {hist['recent_food']}")

    # Final summary
    print("\n" + "="*60)
    print("FINAL STATE")
    print("="*60)
    print(f"\nScientist:")
    print(f"  Remaining Food Supply: {scientist.food_supply}")
    print(f"  Tool Available: {scientist.tool_available}")
    
    print(f"\nPenguins:")
    total_penguin_food = 0
    for penguin in penguins:
        hist = check_history(penguin.name)
        total_penguin_food += penguin.food
        print(f"  {penguin.name}:")
        print(f"    Current Food: {penguin.food}")
        print(f"    Has Tool: {penguin.has_tool}")
        print(f"    Total Received: {hist['recent_food']}")
    
    print(f"\nTotal Food in System:")
    print(f"  Scientist: {scientist.food_supply}")
    print(f"  Penguins: {total_penguin_food}")
    print(f"  Total: {scientist.food_supply + total_penguin_food}")
    
    print("\n" + "="*60)
    print("Simulation Complete!")
    print("="*60)

print("run_simulation function defined!")
````

````python
class PenguinAgent(ToolCallingAgent):
    """Agent representing a penguin that takes actions to find food or request resources."""
    
    def __init__(self, name: str) -> None:
        # Register the find_food tool with this penguin agent
        super().__init__(
            tools=[find_food],
            model=model,
            name=name,
        )
        self.name = name
        self.food = 5  # Start with some food
        self.has_tool = False  # Start without a tool

    def take_action(self) -> Dict[str, Any]:
        """Penguin decides on an action each round."""
        history = check_history(self.name)

        # Decision logic based on current state
        if self.has_tool:
            # If penguin has a tool, prefer fishing (more productive)
            action = "find_food"
            method = "fishing"
            details = "Using tool for efficient fishing"
        elif self.food < 3:
            # If very hungry, request food from scientist
            action = "request_food"
            method = "appeal"
            details = "Low on food, appealing to scientist"
        elif random.random() < 0.6:
            # Otherwise, try foraging (less productive but doesn't need tool)
            action = "find_food"
            method = "foraging"
            details = "Foraging for food"
        else:
            # Sometimes just wait or rest
            action = "rest"
            method = "waiting"
            details = "Taking a break"

        return {
            "action": action,
            "method": method,
            "details": details,
            "food_level": self.food,
            "has_tool": self.has_tool
        }

print("PenguinAgent class defined!")
````

````python
class ScientistAgent(ToolCallingAgent):
    """Agent representing the scientist distributing resources."""
    
    def __init__(self, initial_food_supply: int = 20, refresh_interval: int = 5) -> None:
        super().__init__(
            tools=[check_history, record_distribution],
            model=model,
            name="scientist",
        )
        self.initial_food_supply = initial_food_supply
        self.food_supply = initial_food_supply
        self.tool_available = True
        self.refresh_interval = refresh_interval
        self.turn_counter = 0

    def refresh_resources(self):
        """Periodically refresh the scientist's food supply."""
        self.food_supply = self.initial_food_supply
        self.tool_available = True
        print("\n🔄 Scientist Resources Refreshed!")
        print(f"   Food Supply Reset to: {self.food_supply}")
        print(f"   Tool Availability Reset to: {self.tool_available}")

    def respond_to_action(self, penguin: 'PenguinAgent', penguin_action: Dict[str, Any]) -> None:
        """Respond to a penguin's action by allocating resources."""
        self.turn_counter += 1
        if self.turn_counter % self.refresh_interval == 0:
            self.refresh_resources()

        print(f"\n   [Scientist Responds to {penguin.name}]")
        print(f"   Action: {penguin_action}")
        print(f"   Penguin Food: {penguin.food}, Has Tool: {penguin.has_tool}")

        history = check_history(penguin.name)
        print(f"   History - Recent Food: {history['recent_food']}, Has Tool: {history['has_tool']}")
        print(f"   Scientist Resources - Food: {self.food_supply}, Tool Available: {self.tool_available}")

        # Simple decision-making logic based on penguin state
        try:
            # Decide how much food to give (0-5) based on penguin's food level
            if penguin.food < 3:
                food_to_give = min(5, self.food_supply)  # Generous to hungry penguins
            elif penguin.food < 8:
                food_to_give = min(3, self.food_supply)  # Moderate amount
            else:
                food_to_give = min(1, self.food_supply)  # Minimal to well-fed penguins

            # Decide on tool allocation based on history and availability
            tool_to_give = self.tool_available and penguin.food < 5 and not history['has_tool']

            print(f"   Decision - Food to Give: {food_to_give}, Tool to Give: {tool_to_give}")

            # Apply decisions
            if food_to_give > 0:
                self.food_supply -= food_to_give
                penguin.food += food_to_give

            if tool_to_give:
                penguin.has_tool = True
                self.tool_available = False

            # Record the distribution
            record_distribution(penguin.name, food_to_give, tool_to_give)

            print(f"   Updated - Scientist Food: {self.food_supply}, Penguin Food: {penguin.food}")

        except Exception as e:
            print(f"   Error in scientist response: {e}")

print("ScientistAgent class defined!")
````

````python
# Define Tools

@tool
def find_food(penguin_name: str, method: str) -> int:
    """Find food using a given method.
    
    Args:
        penguin_name: Name of the penguin searching for food
        method: Either 'fishing' (returns 2-7 food) or 'foraging' (returns 0-3 food)
    
    Returns:
        Amount of food found (int)
    """
    if method.lower() == "fishing":
        food_amount = random.randint(2, 7)
        print(f"  🐧 {penguin_name} fished and found {food_amount} food!")
    else:
        food_amount = random.randint(0, 3)
        print(f"  🐧 {penguin_name} foraged and found {food_amount} food!")
    
    return food_amount

@tool
def check_history(penguin_name: str) -> Dict[str, Any]:
    """Check the recent resource distribution history for a specific penguin."""
    history = DISTRIBUTION_HISTORY.get(penguin_name, [])
    recent_food = sum(h["food"] for h in history[-3:]) if history else 0
    has_tool = any(h["has_tool"] for h in history) if history else False
    return {"recent_food": recent_food, "has_tool": has_tool}

@tool
def record_distribution(penguin_name: str, food: int, has_tool: bool) -> str:
    """Record the distribution of resources to a penguin."""
    if penguin_name not in DISTRIBUTION_HISTORY:
        DISTRIBUTION_HISTORY[penguin_name] = []
    DISTRIBUTION_HISTORY[penguin_name].append({"food": food, "has_tool": has_tool})
    return f"Recorded: {penguin_name} got {food} food and {'a' if has_tool else 'no'} tool"

print("Tools defined successfully!")
````
